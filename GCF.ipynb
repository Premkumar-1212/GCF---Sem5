```r
#DECISION TREE
# Install required package
#rpart is for partisation of dataset
install.packages("rpart")

library(rpart)

students <- data.frame(
  StudyHours = c(2, 3, 4, 5, 6, 7, 8, 9, 10),
  Attendance = c(60, 65, 70, 75, 80, 82, 85, 90, 95),
  Result = c(
    "Fail", "Fail", "Fail",
    "Pass", "Pass", "Pass",
    "Pass", "Pass", "Pass"
  )
)

print(students)

# CONVERT RESULT INTO FACTOR
students$Result <- as.factor(students$Result)

# CREATE DECISION TREE MODEL
# Result is the target/output variable.
# StudyHours and Attendance are input variables.

model <- rpart(
  Result ~ StudyHours + Attendance,
  data = students,
  method = "class"
)

print(model)


#PREDICT NEW STUDENT
new_student <- data.frame(
  StudyHours = 6,
  Attendance = 80
)

# Predict whether the student will Pass or Fail
prediction <- predict(
  model,
  new_student,
  type = "class"
)

print(prediction)
```

```r
#k-nearest neighbors in R
#install.packages("class")
#library(class)

StudyHours <- c(1,2,3,4,5,6,7,8)
Marks <- c(30,35,40,50,55,65,75,85)

Result <- c(
  "Fail","Fail","Fail","Pass",
  "Pass","Pass","Pass","Pass"
)

students <- data.frame(
  StudyHours,
  Marks,
  Result
)

#Seprate input and output
#KNN uses StuDyHours And marks to make prediction
X <- students[,c("StudyHours","Marks")]

#this is what we want to predict
Y <- students$Result

new_student <- data.frame(
  StudyHours = 5,
  Marks = 60
)

#Apply KNN
prediction <- knn(
  train = X,            #Training input data
  test = new_student,   #New data to predict
  cl = Y,               #actual classes/results
  k = 3                 #number of nearest neighbors
)

print(prediction)
```

```r
# K-MEANS CLUSTERING IN R
students <- data.frame(
  StudyHours = c(1, 2, 2, 3, 4, 5, 6, 7, 8, 9),
  Marks = c(35, 40, 42, 48, 55, 60, 68, 75, 82, 90)
)

print(students)

# Select the columns that we want to use
# K-Means works with numerical data

data_for_clustering <- students[, c("StudyHours", "Marks")]


# Apply K-Means
# centers = 3 means we want 3 clusters
set.seed(123)  # Makes the result reproducible
kmeans_model <- kmeans(data_for_clustering, centers = 3)

print(kmeans_model$cluster)

# Add the cluster number to our original data
students$Cluster <- kmeans_model$cluster
print(students)


# Display the center of each cluster
# These are called centroids
print(kmeans_model$centers)

# Visualize the clusters

plot(
  students$StudyHours,
  students$Marks,
  col = students$Cluster,       # Different color for each cluster
  pch = 19,                     # Solid circle
  xlab = "Study Hours",
  ylab = "Marks",
  main = "K-Means Clustering"
)

# Add cluster centers to the graph
points(
  kmeans_model$centers[, "StudyHours"],
  kmeans_model$centers[, "Marks"],
  pch = 8,
  cex = 2
)
```

```r
#HIRARCHICAL CLUSTERING IN R
students <- data.frame(
  StuDyHours = c(1,2,2,3,7,8,9,10),
  Marks = c(35,40,42,45,70,75,85,90)
)

print(students)

#calculate distance between students
distance <- dist(students)

print(distance)

#perform hierarchical clustering
hc <- hclust(distance,method = "complete")

print(hc)

#draw the dendrogram
plot(
  hc,
  main = "Hierarchical Clustering Dendrogram",
  xlab = "Students",
  ylab = "Distance"
)

clusters <- cutree(hc, k = 2)

students$Cluster <- clusters

print(students)
```
